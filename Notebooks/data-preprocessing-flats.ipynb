```python
## Rename col "area"
df.rename(columns={'area':'price_per_sqft'}, inplace=True)
```

```python
## preprocessing Society col

print(df['society'].value_counts())
print(df['society'].value_counts().shape)  ## 637 unique socities
```

```python
import re
df['society'] = df['society'].apply(lambda name: re.sub(r'\d+(\.\d+)?\s?★', '', str(name)).strip()).str.lower()
```

```python
df['society'].value_counts()
```

```python
## preprocessing price col
print(df['price'].value_counts())
print(df['price'].value_counts().shape)  ## 440 unique prices
```

```python
df[df['price'] == 'Price on Request']
```

```python
df = df[df['price'] != 'Price on Request']  ## keeping only the rows which have exact prices 
df[df['price'] == 'Price on Request']
```

```python
## standardizing the prices in Cr
def clean_price(x):
    if type(x) == float:
        return x
    if len(x)<2:
        return None
    
    if x[1] == 'Lac':
        return float(x[0])/100
    else:
        return float(x[0])
```

```python
df['price']= df['price'].str.split(' ').apply(clean_price)
```

```python
df.head()
```

```python
print(df['price'].value_counts())
print(df['price'].value_counts().shape)
```

```python
df['price_per_sqft'].value_counts().shape
```

```python
df = df[~(df.astype(str) == df.columns).any(axis=1)]
```

```python
df['price_per_sqft'] = df['price_per_sqft'].str.split('/').str.get(0).str.replace('₹','').str.replace(',','').str.strip().astype('float')
```
